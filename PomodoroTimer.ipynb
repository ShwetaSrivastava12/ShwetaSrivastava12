```python
import ttkbootstrap as tb
from ttkbootstrap.constants import *
from tkinter import messagebox
import time

class PomodoroTimer:
    def __init__(self, root):
        self.root = root
        self.root.title("Pomodoro Timer")

        self.work_time = 25 * 60  # 25 minutes
        self.short_break = 5 * 60
        self.long_break = 15 * 60
        self.sessions = 0
        self.is_running = False
        self.remaining_time = self.work_time

        self.label = tb.Label(root, text="Pomodoro Timer", font=("Helvetica", 24))
        self.label.pack(pady=20)

        self.timer_label = tb.Label(root, text="", font=("Helvetica", 48))
        self.timer_label.pack(pady=10)

        self.start_btn = tb.Button(root, text="Start", command=self.start_timer, bootstyle="success")
        self.start_btn.pack(side="left", padx=20)

        self.reset_btn = tb.Button(root, text="Reset", command=self.reset_timer, bootstyle="danger")
        self.reset_btn.pack(side="right", padx=20)

        self.update_timer_display()

    def update_timer_display(self):
        minutes = self.remaining_time // 60
        seconds = self.remaining_time % 60
        self.timer_label.config(text=f"{minutes:02d}:{seconds:02d}")

    def countdown(self):
        if self.is_running and self.remaining_time > 0:
            self.remaining_time -= 1
            self.update_timer_display()
            self.root.after(1000, self.countdown)
        elif self.remaining_time == 0:
            self.sessions += 1
            self.is_running = False
            messagebox.showinfo("Session Complete", "Time's up!")

            if self.sessions % 4 == 0:
                self.remaining_time = self.long_break
            elif self.sessions % 2 == 0:
                self.remaining_time = self.short_break
            else:
                self.remaining_time = self.work_time

            self.update_timer_display()

    def start_timer(self):
        if not self.is_running:
            self.is_running = True
            self.countdown()

    def reset_timer(self):
        self.is_running = False
        self.remaining_time = self.work_time
        self.update_timer_display()

# Run this cell only once
app = tb.Window(themename="superhero")  # or "minty", "darkly", etc.
pomodoro = PomodoroTimer(app)
app.mainloop()
```
